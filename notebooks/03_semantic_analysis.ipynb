```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

sentences = [
    "ChatGPT explains difficult concepts in simple language.",
    "AI makes complicated topics easy to understand.",
    "Teachers provide personal guidance to students."
]

embeddings = model.encode(sentences)
print(embeddings.shape)
print(model.similarity(embeddings, embeddings))
```

```python
import pandas as pd
df = pd.read_csv("data/synthetic/synthetic_responses_cleaned.csv")
print(df.columns.tolist())
df.head()
```

```python
print(df.columns.tolist())
print(df.iloc[0].to_dict())
```

```python
import pandas as pd

raw = df.copy()   # df is the cleaned file you already loaded; we leave it untouched

work = pd.DataFrame({
    "respondent_id": ["R" + str(i + 1).zfill(3) for i in range(len(raw))],
    "preference_overall": raw["16. Overall, which source do you prefer for academic help?"],
    "reason_llm": raw["23. What are the main reasons you prefer ChatGPT/LLMs for academic help?"],
    "reason_teacher": raw["24. What are the main reasons you prefer teachers for academic help?"],
    "reason_overall": raw["27. Why do you prefer ChatGPT/LLMs or teachers for academic help?"],
    "llm_situations": raw["28. In what situations do you think ChatGPT/LLMs are more useful than teachers, and why?"],
    "teacher_situations": raw["29. In what situations do you think teachers are more useful than ChatGPT/LLMs, and why?"],
})

records = []
for _, row in work.iterrows():
    # Q23, Q24: multi-select answers, so split into separate phrases
    for col in ["reason_llm", "reason_teacher"]:
        if pd.notna(row[col]):
            for phrase in str(row[col]).split(","):
                if phrase.strip():
                    records.append({"respondent_id": row["respondent_id"],
                                    "question": col,
                                    "text": phrase.strip()})
    # Q27, Q28, Q29: free text, keep each answer whole
    for col in ["reason_overall", "llm_situations", "teacher_situations"]:
        if pd.notna(row[col]) and str(row[col]).strip():
            records.append({"respondent_id": row["respondent_id"],
                            "question": col,
                            "text": str(row[col]).strip()})

texts_df = pd.DataFrame(records)
print(texts_df.shape)
print(texts_df["question"].value_counts())
print("Unique texts:", texts_df["text"].nunique())
texts_df.head(10)
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np

model = SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")

unique_texts = texts_df["text"].unique().tolist()
unique_emb = model.encode(unique_texts, convert_to_numpy=True,
                          normalize_embeddings=True, show_progress_bar=True)

# map each of the 1,936 rows to its vector
text_to_idx = {t: i for i, t in enumerate(unique_texts)}
embeddings = unique_emb[[text_to_idx[t] for t in texts_df["text"]]]

print(unique_emb.shape)   # expect (28, 384)
print(embeddings.shape)   # expect (1936, 384)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

sim = cosine_similarity(unique_emb)
i_idx, j_idx = np.triu_indices(len(unique_texts), k=1)

pairs_df = pd.DataFrame({
    "text_1": [unique_texts[i] for i in i_idx],
    "text_2": [unique_texts[j] for j in j_idx],
    "similarity": sim[i_idx, j_idx]
}).sort_values("similarity", ascending=False)

pd.set_option("display.max_colwidth", 70)
pd.set_option("display.width", 200)
print(pairs_df.head(10).to_string(index=False))
print(pairs_df.tail(10).to_string(index=False))
```

```python
import os
os.makedirs("results/semantic", exist_ok=True)

texts_df.to_csv("results/semantic/text_responses.csv", index=False)
pairs_df.to_csv("results/semantic/similarity_pairs.csv", index=False)
np.save("results/semantic/unique_embeddings.npy", unique_emb)

!ls results/semantic
```
