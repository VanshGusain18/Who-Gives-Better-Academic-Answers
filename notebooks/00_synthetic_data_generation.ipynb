
# Synthetic Data Generation

## Project
Who Explains Better? A Comparative Study of Student Perceptions of Teacher-Given and LLM-Generated Academic Explanations.

## Purpose
This notebook generates synthetic survey responses for testing and validating the project's data processing, NLP, semantic analysis, and statistical analysis pipeline.

## Dataset
- 200 synthetic respondents
- 40 columns matching the Google Forms survey structure
- Likert-scale responses
- Categorical demographic responses
- Synthetic open-ended responses

## Important Note
This dataset is generated exclusively for development and testing purposes. It must not be used as actual research data or reported as real participant findings.

## Output
`data/synthetic/synthetic_responses.csv`


```python

import os
import random
import pandas as pd
from datetime import datetime, timedelta

# --------------------------------------------------
# 1. CONFIGURATION
# --------------------------------------------------

random.seed(42)

BASE_DIR = "/content/Who-Gives-Better-Academic-Answers"
OUTPUT_DIR = os.path.join(BASE_DIR, "data", "synthetic")
OUTPUT_FILE = os.path.join(OUTPUT_DIR, "synthetic_responses.csv")

os.makedirs(OUTPUT_DIR, exist_ok=True)

N_RESPONDENTS = 200

# --------------------------------------------------
# 2. EXACT GOOGLE FORMS COLUMN HEADERS
# --------------------------------------------------

columns = [
    "Timestamp",
    "1. Do you agree to participate in this research survey?",
    "2. Are you 18 years of age or older?",
    "3. What is your program and current semester/year?",
    "4. How often do you use ChatGPT or other LLMs for academic study?",
    "5. How often do you ask teachers for academic help?",
    "6. How familiar are you with ChatGPT and other LLMs?",
    "7. ChatGPT/LLM answers are usually accurate for my academic subjects.",
    "8. Teachers understand my course requirements and academic context better.",
    "9. ChatGPT/LLMs provide academic answers faster than teachers.",
    "10. Teacher explanations help me understand the reasoning behind an answer, not just the final answer.",
    "11. ChatGPT/LLM explanations are clear and easy to understand.",
    "12. I trust teachers more when the academic task is important, such as an exam or major project.",
    "13. ChatGPT/LLMs make it easier for me to understand difficult academic concepts.",
    "14. Teachers provide useful guidance when I have follow-up questions or doubts.",
    "15. ChatGPT/LLMs can adapt explanations to my level and learning needs.",
    "16. Overall, which source do you prefer for academic help?",
    "17. Concept explanation — For example: What is Backpropagation?",
    "18. Step-by-step problem solving — For example: How do I approach a programming problem?",
    "19. Numerical or technical problem solving — For example: Calculate or interpret a technical metric.",
    "20. Course-specific questions — For example: Questions related to your syllabus, course, or institution.",
    "21. Writing improvement — For example: Improving clarity, structure, grammar, or tone.",
    "22. Follow-up questions and doubts — For example: Asking multiple questions to clarify a concept.",
    "23. What are the main reasons you prefer ChatGPT/LLMs for academic help?",
    "24. What are the main reasons you prefer teachers for academic help?",
]

q25 = [
    "25. Rate ChatGPT/LLMs on the following aspects. [Accuracy]",
    "25. Rate ChatGPT/LLMs on the following aspects. [Clarity of explanation]",
    "25. Rate ChatGPT/LLMs on the following aspects. [Helpfulness for understanding concepts]",
    "25. Rate ChatGPT/LLMs on the following aspects. [Practical usefulness]",
    "25. Rate ChatGPT/LLMs on the following aspects. [Ease of understanding]",
    "25. Rate ChatGPT/LLMs on the following aspects. [Trustworthiness]",
]

q26 = [
    "26. Rate teachers on the following aspects. [Accuracy]",
    "26. Rate teachers on the following aspects. [Clarity of explanation]",
    "26. Rate teachers on the following aspects. [Helpfulness for understanding concepts]",
    "26. Rate teachers on the following aspects. [Practical usefulness]",
    "26. Rate teachers on the following aspects. [Ease of understanding]",
    "26. Rate teachers on the following aspects. [Trustworthiness]",
]

q27 = "27. Why do you prefer ChatGPT/LLMs or teachers for academic help?"
q28 = "28. In what situations do you think ChatGPT/LLMs are more useful than teachers, and why?"
q29 = "29. In what situations do you think teachers are more useful than ChatGPT/LLMs, and why?"

columns.extend(q25)
columns.extend(q26)
columns.extend([q27, q28, q29])

assert len(columns) == 40, f"Expected 40 columns, got {len(columns)}"

# --------------------------------------------------
# 3. RESPONSE OPTIONS
# --------------------------------------------------

programs = [
    "B.Tech CSE, 7th Semester",
    "B.Tech CSE, 5th Semester",
    "B.Tech IT, 7th Semester",
    "B.Tech ECE, 5th Semester",
    "B.Tech CSE, 3rd Semester",
    "B.Tech Mechanical, 5th Semester",
]

llm_frequency = [
    "Daily",
    "Several times a week",
    "Once a week",
    "Rarely",
]

teacher_frequency = [
    "Daily",
    "Several times a week",
    "Once a week",
    "Rarely",
    "Never",
]

familiarity = [1, 2, 3, 4, 5, 6, 7]

preferences = [
    "ChatGPT / LLM",
    "Teacher",
    "Both are about the same",
    "Depends on the situation",
]

use_cases = [
    "ChatGPT / LLM",
    "Teacher",
    "Both are about the same",
    "Depends on the situation",
]

llm_reasons = [
    "Faster responses",
    "Available anytime",
    "Clear step-by-step explanations",
    "Easy-to-understand language",
    "More examples",
    "Can explain the same concept in different ways",
    "Helps with difficult concepts",
    "Personalized responses",
]

teacher_reasons = [
    "Better understanding of course/syllabus",
    "Better follow-up and doubt solving",
    "Practical/real-world experience",
    "More trustworthy for important academic decisions",
]

llm_texts = [
    "I prefer ChatGPT because it is available anytime and explains concepts step by step.",
    "LLMs provide examples and help me understand difficult topics quickly.",
    "ChatGPT is useful for personalized explanations and repeated follow-up questions.",
    "I prefer LLMs for quick explanations, coding help, and understanding theoretical concepts.",
]

teacher_texts = [
    "Teachers understand the syllabus and provide better course-specific guidance.",
    "Teachers are more useful for clearing doubts and providing practical examples.",
    "Teachers have experience and can guide us during exams and important projects.",
    "Teachers provide reliable feedback and understand our academic requirements.",
]

# --------------------------------------------------
# 4. GENERATE SYNTHETIC RESPONSES
# --------------------------------------------------

rows = []

for i in range(N_RESPONDENTS):

    profile = random.choice(preferences)

    # Ratings are consistently on a 1-7 scale.
    llm_rating = random.randint(4, 7)
    teacher_rating = random.randint(4, 7)

    # Create a slightly different preference pattern.
    if profile == "ChatGPT / LLM":
        llm_rating = random.randint(5, 7)
        teacher_rating = random.randint(3, 6)

    elif profile == "Teacher":
        llm_rating = random.randint(3, 6)
        teacher_rating = random.randint(5, 7)

    elif profile == "Both are about the same":
        llm_rating = random.randint(4, 6)
        teacher_rating = random.randint(4, 6)

    else:
        llm_rating = random.randint(3, 7)
        teacher_rating = random.randint(3, 7)

    # Timestamp
    timestamp = datetime(2026, 11, 1) + timedelta(
        minutes=random.randint(0, 60 * 24 * 14)
    )

    row = {
        columns[0]: timestamp.strftime("%d/%m/%Y %H:%M:%S"),
        columns[1]: "I agree",
        columns[2]: "Yes",
        columns[3]: random.choice(programs),
        columns[4]: random.choice(llm_frequency),
        columns[5]: random.choice(teacher_frequency),
        columns[6]: random.choice(familiarity),
    }

    # Q7-Q15: academic perception statements
    for col in columns[7:16]:
        row[col] = random.randint(1, 7)

    # Q16: overall preference
    row[columns[16]] = profile

    # Q17-Q22: use-case preferences
    for col in columns[17:23]:
        row[col] = random.choice(use_cases)

    # Q23-Q24: reasons
    row[columns[23]] = ", ".join(
        random.sample(llm_reasons, random.randint(2, 5))
    )

    row[columns[24]] = ", ".join(
        random.sample(teacher_reasons, random.randint(2, 4))
    )

    # Q25: ChatGPT ratings
    for col in q25:
        row[col] = random.randint(1, 7)

    # Q26: Teacher ratings
    for col in q26:
        row[col] = random.randint(1, 7)

    # Q27-Q29: open-ended responses
    row[q27] = random.choice(llm_texts + teacher_texts)
    row[q28] = random.choice([
        "When I need quick explanations or have limited time.",
        "For difficult concepts, coding questions, and repeated doubts.",
        "When I need examples or personalized explanations.",
        "For quick revision and understanding theoretical topics.",
    ])
    row[q29] = random.choice([
        "Teachers are better for syllabus-specific and exam-oriented guidance.",
        "When I need practical experience or detailed feedback.",
        "For important academic decisions and clearing complex doubts.",
        "When personal guidance and classroom interaction are needed.",
    ])

    rows.append(row)

# --------------------------------------------------
# 5. SAVE AND VERIFY
# --------------------------------------------------

df = pd.DataFrame(rows, columns=columns)

df.to_csv(OUTPUT_FILE, index=False, encoding="utf-8-sig")

print("Synthetic RAW dataset generated successfully!")
print("Shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Saved at:", OUTPUT_FILE)

display(df.head())
```
