```python

import os
import random
import pandas as pd
from datetime import datetime, timedelta

# Reproducibility
random.seed(42)

# Repository paths
BASE_DIR = "/content/Who-Gives-Better-Academic-Answers"
OUTPUT_DIR = os.path.join(BASE_DIR, "data", "synthetic")
OUTPUT_FILE = os.path.join(OUTPUT_DIR, "synthetic_responses.csv")

os.makedirs(OUTPUT_DIR, exist_ok=True)

# --------------------------------------------------
# 1. Dataset column names (40 columns)
# --------------------------------------------------

columns = [
    "Timestamp",
    "1. Consent",
    "2. Age eligibility",
    "3. Program and semester",
    "4. Frequency of LLM usage",
    "5. Frequency of teacher help",
    "6. Familiarity with LLMs",
    "7. Overall preference",
    "8. LLM usefulness",
    "9. Teacher usefulness",
    "10. LLM understanding",
    "11. Teacher understanding",
    "12. LLM explanation clarity",
    "13. Teacher explanation clarity",
    "14. LLM examples",
    "15. Teacher examples",
    "16. Overall preference",
    "17. Concept explanation",
    "18. Step-by-step explanation",
    "19. Numerical problem solving",
    "20. Course-specific questions",
    "21. Writing improvement",
    "22. Follow-up questions",
    "23. Reasons for using LLMs",
    "24. Reasons for preferring teachers",
]

# Q25: LLM ratings
q25 = [
    "25. Rate ChatGPT/LLMs on the following aspects. [Accuracy]",
    "25. Rate ChatGPT/LLMs on the following aspects. [Clarity of explanation]",
    "25. Rate ChatGPT/LLMs on the following aspects. [Helpfulness for understanding concepts]",
    "25. Rate ChatGPT/LLMs on the following aspects. [Practical usefulness]",
    "25. Rate ChatGPT/LLMs on the following aspects. [Ease of understanding]",
    "25. Rate ChatGPT/LLMs on the following aspects. [Trustworthiness]",
]

# Q26: Teacher ratings
q26 = [
    "26. Rate teachers on the following aspects. [Accuracy]",
    "26. Rate teachers on the following aspects. [Clarity of explanation]",
    "26. Rate teachers on the following aspects. [Helpfulness for understanding concepts]",
    "26. Rate teachers on the following aspects. [Practical usefulness]",
    "26. Rate teachers on the following aspects. [Ease of understanding]",
    "26. Rate teachers on the following aspects. [Trustworthiness]",
]

columns += q25 + q26 + [
    "27. Explain your preference",
    "28. How are LLMs useful?",
    "29. How are teachers useful?",
]

assert len(columns) == 40, f"Expected 40 columns, got {len(columns)}"

# --------------------------------------------------
# 2. Synthetic response options
# --------------------------------------------------

programs = [
    "B.Tech CSE - Semester 1",
    "B.Tech CSE - Semester 3",
    "B.Tech CSE - Semester 5",
    "B.Tech CSE - Semester 7",
    "B.Tech IT - Semester 3",
    "B.Tech ECE - Semester 5",
    "B.Tech Other - Semester 5",
]

llm_frequency = [
    "Daily", "Several times a week",
    "Weekly", "Rarely", "Never"
]

teacher_frequency = [
    "Daily", "Several times a week",
    "Weekly", "Rarely", "Never"
]

familiarity = [
    "Very familiar", "Familiar",
    "Somewhat familiar", "Slightly familiar"
]

preference = [
    "Teachers", "ChatGPT/LLMs",
    "Both equally", "Depends on the situation"
]

likert = [1, 2, 3, 4, 5]

llm_reasons = [
    "Quick responses, availability, easy explanations",
    "Step-by-step explanations and examples",
    "Useful for revision and doubt solving",
    "Can ask follow-up questions",
    "Helps simplify difficult concepts"
]

teacher_reasons = [
    "Teachers understand our course requirements",
    "Teachers provide accurate explanations",
    "Teachers clarify doubts interactively",
    "Teachers provide practical examples",
    "Teachers understand student difficulties"
]

# --------------------------------------------------
# 3. Generate 200 synthetic respondents
# --------------------------------------------------

rows = []

start_time = datetime(2026, 9, 1, 9, 0)

for i in range(200):

    timestamp = start_time + timedelta(
        minutes=random.randint(1, 50000)
    )

    profile = random.choice(preference)

    # Ratings: synthetic patterns, not real research findings
    if profile == "ChatGPT/LLMs":
        llm_ratings = [random.choices(
            likert, weights=[2, 5, 15, 40, 38])[0]
            for _ in range(6)
        ]
        teacher_ratings = [random.choices(
            likert, weights=[5, 15, 35, 30, 15])[0]
            for _ in range(6)
        ]

    elif profile == "Teachers":
        llm_ratings = [random.choices(
            likert, weights=[10, 20, 35, 25, 10])[0]
            for _ in range(6)
        ]
        teacher_ratings = [random.choices(
            likert, weights=[2, 5, 15, 40, 38])[0]
            for _ in range(6)
        ]

    else:
        llm_ratings = [random.choice(likert) for _ in range(6)]
        teacher_ratings = [random.choice(likert) for _ in range(6)]

    row = [
        timestamp.strftime("%d/%m/%Y %H:%M:%S"),
        "Yes",
        "Yes",
        random.choice(programs),
        random.choice(llm_frequency),
        random.choice(teacher_frequency),
        random.choice(familiarity),
        profile,
        random.choice(preference),
        random.choice(preference),
        random.choice(preference),
        random.choice(preference),
        random.choice(preference),
        random.choice(preference),
        random.choice(preference),
        random.choice(preference),
        profile,
        random.choice(preference),
        random.choice(preference),
        random.choice(preference),
        random.choice(preference),
        random.choice(preference),
        random.choice(preference),
        random.choice(llm_reasons),
        random.choice(teacher_reasons),
    ]

    row += llm_ratings
    row += teacher_ratings

    row += [
        random.choice([
            "I prefer clear and detailed explanations.",
            "It depends on the topic and situation.",
            "Both sources have different strengths.",
            "Teachers provide guidance while LLMs are convenient."
        ]),
        random.choice([
            "LLMs help with quick explanations and revision.",
            "They simplify complex concepts.",
            "They help with follow-up questions."
        ]),
        random.choice([
            "Teachers clarify doubts and guide students.",
            "Teachers provide practical examples.",
            "Teachers understand course expectations."
        ])
    ]

    rows.append(row)

# --------------------------------------------------
# 4. Create and save raw dataset
# --------------------------------------------------

df = pd.DataFrame(rows, columns=columns)

df.to_csv(OUTPUT_FILE, index=False)

print("Synthetic RAW dataset generated successfully!")
print("Shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Saved at:", OUTPUT_FILE)

display(df.head())
```
