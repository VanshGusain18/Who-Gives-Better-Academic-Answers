```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

sentences = [
    "ChatGPT explains difficult concepts in simple language.",
    "AI makes complicated topics easy to understand.",
    "Teachers provide personal guidance to students."
]

embeddings = model.encode(sentences)
print(embeddings.shape)
print(model.similarity(embeddings, embeddings))
```

```python
sentences = [
    "ChatGPT explains difficult concepts in simple language.",
    "ChatGPT mushkil concepts ko aasan bhasha me samjhata hai.",   # Hinglish
    "चैटजीपीटी कठिन अवधारणाओं को सरल भाषा में समझाता है।",          # Hindi
    "Teachers provide personal guidance to students."
]
embeddings = model.encode(sentences)
print(model.similarity(embeddings, embeddings))
```
